```python
import numpy as np
import pandas as pd
```

```python
train = pd.read_csv('train.csv')
train = train.dropna(subset=['x', 'y'])
X_train = train['x'].values.reshape(-1,1)
Y_train = train['y'].values.reshape(-1,1)
```

```python
#scaling features
min = np.min(X_train)
max = np.max(X_train)
ymin = np.min(Y_train)
ymax = np.max(Y_train)
X_train = [(X - min)/(max - min) for X in X_train]
Y_train = [(Y- ymin)/(ymax - ymin) for Y in Y_train]
```

```python
m = len(X_train)
X_train_new = np.c_[np.ones((m, 1)), X_train]

lr = 0.02
epochs = 10000
theta = np.zeros((2, 1))
for epoch in range(epochs):
  Y_pred = np.matmul(X_train_new, theta)
  gradients = (2/m) * (np.matmul(X_train_new.T, (Y_pred - Y_train)))
  theta = theta - lr * gradients

print("Intercept: ", theta[0][0])
print("Slope: ", theta[1][0])
```
